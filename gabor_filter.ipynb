```python
source_base = Path("./dataset_unprocessed/FingerVein/001/Finger Vein Database/Finger Vein Database")
dest_base = Path("./dataset_organized/Finger_Vein_Database/")

Path("./dataset_organized/Finger_Vein_Database").mkdir(exist_ok=True)

for dataset_folder in source_base.iterdir():
    #print(dataset_folder)
    if dataset_folder.is_dir():
        dataset_num = dataset_folder.name  # This gives you 001, 002, 003, etc.
        #print(dataset_num)
        
        for bmp_file in dataset_folder.rglob("*.bmp"):
            # Determine right/left from path
            path_lower = str(bmp_file).lower()
            hand_code = "r" if "right" in path_lower else "l"
            
            # Create new filename
            new_name = f"{dataset_num}_{hand_code}_{bmp_file.name}"
            #print(new_name)
            
            # Copy file
            shutil.copy2(bmp_file, dest_base / new_name)
        #     break
        # break
print("Organization complete!")
```

```python
dataset_name = "FV_USM"
organized_path = Path(f"./dataset_organized/{dataset_name}/")

print("Renaming files while preserving uniqueness...")

# Initialize a counter for the new filenames
counter = 1
last_prefix = None

for jpg_file in organized_path.glob("*.jpg"):
    current_name = jpg_file.stem
    
    # Split the filename into parts
    parts = current_name.split('_')
    
    if len(parts) == 3:
        prefix, middle, suffix = parts
        
        # If the prefix has changed, reset the counter
        if prefix != last_prefix:
            counter = 1
            last_prefix = prefix
        
        # Construct the new name: prefix and incremented counter
        new_name = f"{prefix}_{str(counter).zfill(2)}"  # Format the number as 01, 02, 03...
        new_filename = f"{new_name}.jpg"
        new_filepath = organized_path / new_filename
        
        # Rename the file
        jpg_file.rename(new_filepath)
        #print(f"Renamed: {jpg_file.name} -> {new_filename}")

        # Increment the counter for the next file
        counter += 1

print("Done!")
```

```python
TARGET_SIZE = (320, 240)  # (width, height)
dataset_name = "FV_USM"
path = Path(f"./dataset_organized/{dataset_name}/")

path.mkdir(parents=True, exist_ok=True)

def resize_with_crop(img, target_size):
    h, w = img.shape[:2]
    target_w, target_h = target_size
    
    # Calculate scaling factor (scale to cover the entire target area)
    scale = max(target_w / w, target_h / h)
    
    # Resize image
    new_w = int(w * scale)
    new_h = int(h * scale)
    resized = cv2.resize(img, (new_w, new_h), interpolation=cv2.INTER_AREA)
    
    # Crop center to target size
    start_x = (new_w - target_w) // 2
    start_y = (new_h - target_h) // 2
    cropped = resized[start_y:start_y+target_h, start_x:start_x+target_w]
    
    return cropped

for filepath in path.glob("*.jpg"):
    img = cv2.imread(str(filepath), cv2.IMREAD_GRAYSCALE)
    
    # Resize with crop (maintains aspect ratio)
    resized = resize_with_crop(img, TARGET_SIZE)
    
    # Save the resized image
    cv2.imwrite(str(filepath), resized)

dataset_name = "MMCBNU_6000"
path = Path(f"./dataset_organized/{dataset_name}/")

path.mkdir(parents=True, exist_ok=True)

for filepath in path.glob("*.jpg"):
    img = cv2.imread(str(filepath), cv2.IMREAD_GRAYSCALE)
    
    # Resize with crop (maintains aspect ratio)
    resized = resize_with_crop(img, TARGET_SIZE)
    
    # Save the resized image
    cv2.imwrite(str(filepath), resized)

print("Resizing with crop complete!")
```
