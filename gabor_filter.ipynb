```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
import glob
import os
from pathlib import Path
import shutil
```

```python
source_base = Path("./dataset_unprocessed/FingerVein/001/Finger Vein Database/Finger Vein Database")
dest_base = Path("./dataset_organized/Finger_Vein_Database/")

Path("./dataset_organized/Finger_Vein_Database").mkdir(exist_ok=True)

for dataset_folder in source_base.iterdir():
    #print(dataset_folder)
    if dataset_folder.is_dir():
        dataset_num = dataset_folder.name  # This gives you 001, 002, 003, etc.
        #print(dataset_num)
        print(f"Organizing dataset {dataset_num}...")
        
        for bmp_file in dataset_folder.rglob("*.bmp"):
            # Determine right/left from path
            path_lower = str(bmp_file).lower()
            hand_code = "r" if "right" in path_lower else "l"
            
            # Create new filename
            new_name = f"{dataset_num}_{hand_code}_{bmp_file.name}"
            #print(new_name)
            
            # Copy file
            shutil.copy2(bmp_file, dest_base / new_name)
        #     break
        # break
print("Organization complete!")
```

```python
dataset_name = "MMCBNU_6000"
source_base = Path(f"./dataset_unprocessed/FingerVein/002/{dataset_name}/{dataset_name}/{dataset_name}/Captured images")
dest_base = Path(f"./dataset_organized/{dataset_name}/")

dest_base.mkdir(exist_ok=True)

for dataset_folder in source_base.iterdir():
    #print(dataset_folder)
    if dataset_folder.is_dir():
        dataset_num = dataset_folder.name  # This gives you 001, 002, 003, etc.
        #print(dataset_num)
        print(f"Organizing dataset {dataset_num}...")
        count = 1
        hand_code = 'l'
        for bmp_file in dataset_folder.rglob("*.bmp"):
            # Switch to right hand after 30 images
            if count == 31:
                hand_code = 'r'
                count = 1
            
            # Create new filename
            new_name = f"{dataset_num}_{hand_code}_{count:02d}.bmp"
            #print(new_name)
            
            # Copy file
            shutil.copy2(bmp_file, dest_base / new_name)
            #print(count)
            count = count + 1
        #     break
        #break
print("Organization complete!")
```
